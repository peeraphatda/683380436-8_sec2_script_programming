```python
size = 5
for i in range(size):
    for j in range(size):
        print("*", end="")
    print()
```

```python
rows = 5
for i in range(1, rows + 1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
rows = 5
for i in range(rows, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
rows = 5
for i in range(1, rows + 1):
    for j in range(rows - i):
        print(" ", end="")
    for k in range(2 * i - 1):
        print("*", end="")
    print()
```
